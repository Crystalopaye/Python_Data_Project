# Pandas: Analyzing

```python
import pandas as pd
from datasets import load_dataset

# Loading Data
dataset = load_dataset("lukebarousse/data_jobs")
df = dataset["train"].to_pandas()

# Data Cleanup
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
```

```python
df.describe()
```

```python
# Analyzing the median
df["salary_year_avg"].median()
```

```python
# Analzing the min salary
df["salary_year_avg"].min()
```

```python
# Index value of the minimum salary
min_salary = df["salary_year_avg"].idxmin()
```

```python
# Using iloc to find the integer location of the min salary
df.iloc[min_salary]
```

```python
# Finding the unique values of the job_title_short
df["job_title_short"].unique()
```

```python
# Find the unique value counts for job_title_short
df["job_title_short"].value_counts()
```

### Pandas Basics

```python
df["job_country"].value_counts().head(20)
```

```python
# Find if Ghana is in the Dataset using the .isin() function.
# Using the any() function to check if it exists in any portion of the dataset
df["job_country"].isin(["Ghana"]).any()
```

```python
# Filtering for US jobs
us_jobs = df[df["job_country"] == "United States"]
```

```python
us_jobs
```

```python
# Using the .notna() function to only display non na salaries
us_jobs = us_jobs[us_jobs["salary_year_avg"].notna()]
```

```python
us_jobs
```

```python
# Grouping by the job_title_short and show the salary_year_avg
# Sort the median values by descending order
us_jobs.groupby(by="job_title_short")["salary_year_avg"].agg(["median", "min", "max", "count"]).sort_values(by="median", ascending=False)
```

### Bar Chart - Counts of Job Postings

```python
job_counts = df.job_title_short.value_counts()
```

```python
job_counts
```

```python
# Plot a bar chart with the .bar function
plt.bar(job_counts.index, job_counts)
plt.title('Postings by Job Title')
plt.ylabel("Count of Job Postings")
plt.xticks(rotation=45, ha="right") #rotation=45 angles the x-axes lable 45. ha(horizontal alignment-aligns the label horizontal)
plt.show()
```

```python
# horizontal bar chart
job_counts = job_counts.sort_values(ascending=True)
plt.barh(job_counts.index, job_counts)
plt.show()
```
