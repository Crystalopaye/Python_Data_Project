```python
# First lets create a list with 10 yearly salaries for a Senior Data Analyst job.
# We're just using a combination of the random library to get random integers between 100000 and 150000.
# Then using a for loop to get 10 (random) values.

import random as rd
```

```python
salary = [rd.randint(100000, 150000) for num in range(10)]
salary
```

```python
# Convert the list to a numpy array
salary_array = np.array(salary)
salary_array
```

```python
# Calculate the total sum of salaries
total_sum_salaries = np.sum(salary_array)
total_sum_salaries
```

```python
# Calculate the product of the elements in the array salary_array
prod_salaries = np.prod(salary_array)
prod_salaries
```

```python
# Calculate the cumulative sum of salary_array
cumulative_sum = np.cumsum(salary_array)
cumulative_sum
```

```python
# Calculate the cumulative product of salary_sarry
cumulative_prod = np.cumprod(salary_array)
cumulative_prod
```

### Statistics Operators

```python
# Calculate the mean of salary_array
average_salary = np.mean(salary_array)
average_salary
```

```python
# Calculate the median
median_salary = np.median(salary_array)
median_salary
```

```python
# Calculate the variance
salary_variance = np.var(salary_array, ddof=1)
salary_variance
```

```python
# Calculate the standard deviation
np.std(salary_array, ddof=1)
```

```python
# Calculate the minimum element in salary_array
min_salary = np.min(salary_array)
min_salary
```

```python
# Calculate the maximum element in salary_array
max_salary = np.max(salary_array)
max_salary
```

### NAN

```python
# Inserting Missing Values
salary_with_nan = np.array([12345, np.nan, 23553, 84467, 973783, np.nan], dtype=float )
salary_with_nan
```

```python
# Replace Values with NaN
salary_with_nan[salary_with_nan < 20000] = np.nan
salary_with_nan
```

### Where

```python
# np.where(condition)
salary_array = np.where(salary_array < 120_000, 120_000, salary_array)
salary_array
```

### Random Sampling

```python
# np.random.normal
# Syntax: np.random.normal(loc=0.0, scale=1.0, size=None)

# A few other random sampling functions:
# np.random.rand
# np.random.randn
# np.random.randint
# np.random.random
# np.random.uniform
# np.random.binomial
# np.random.poisson

# Example
# Let's add some random noise to the salary_array to simulate salary variations.
# We can generate random values from a normal distribution with a mean of 0 and a
# standard deviation of 5000, then add these values to the salaries.

noise = np.random.normal(0, 5000, salary_array.size)
salary_array_with_noise = salary_array + noise
salary_array_with_noise
```
