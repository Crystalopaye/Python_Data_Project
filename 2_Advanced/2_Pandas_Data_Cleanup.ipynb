```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset("lukebarousse/data_jobs")
df = dataset["train"].to_pandas()

# Data Cleanup
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
```

```python
# Replacing the na values in the salary_year_avg and salary_hour_avg with median values
# Find the median of the salary_year_avg
median_salary_year = df["salary_year_avg"].median()
```

```python
median_salary_hour = df["salary_hour_avg"].median()
```

```python
# Replace the na values using df.fillna
df_filled = df # to keep the integrity of the original df

df_filled["salary_year_avg"] = df_filled["salary_year_avg"].fillna(median_salary_year)
df["salary_hour_avg"] = df_filled["salary_hour_avg"].fillna(median_salary_hour)
```

```python
# Drop duplicates from the dataset using df.drop_duplicates
df_unique = df_filled # To keep the intergrity of the original data set

df_unique = df_unique.drop_duplicates()

print("Lenght of the original df: ", len(df_filled))
print("Lenght of drop duplicates df: ", len(df_unique))
print("rows droped:                ", len(df_filled)-len(df_unique))
```

```python
# Removing duplicates from the job_title and company_name using the drop_duplicates(subset=()) function
df_unique = df_unique.drop_duplicates(subset=["job_title", "company_name"])

print("Lenght of the original df: ", len(df_filled))
print("Lenght of drop duplicates df: ", len(df_unique))
print("rows droped:                ", len(df_filled)-len(df_unique))
```
