```python
# Import Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Load Data
dataset = load_dataset("lukebarousse/data_jobs")
df = dataset["train"].to_pandas()

# Data Cleanup
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
```

```python
# Creating a fake data to concat. We'll break the months into separate dataframes

# Extract the month from job_posted_Date
df["job_posted_month"] = df["job_posted_date"].dt.strftime("%b")
```

```python
# Breaking down the job_posted_month into separate months using a dictionary
# dict_months = {
#     "Jan": df_jan,
#     "Feb": df_feb,
#     "Mar": df_mar,
#     "Apr": df_apr,
#     "May": df_may,
#     "Jun": df_jun,
#     "Jul": df_jul,
#     "Aug": df_aug,
#     "Sep": df_sep,
#     "Oct": df_oct,
#     "Nov": df_nov,
#     "Dec": df_dec
# }

# extract the unique months from the months column
months = df["job_posted_month"].unique()
```

```python
# Building the dictionary comprehension
dict_months = {month: df[df["job_posted_month"] == month] for month in months}  # Replacing the month with the df_month

dict_months["Jan"]
```

### Boss needs 2023 Quarter 1

```python
df_q1 = pd.concat([dict_months["Jan"], dict_months["Feb"], dict_months["Mar"]], ignore_index=True)
```

```python
df_q1["job_posted_month"].value_counts().plot(kind="bar")
plt.title("First Quarter Job Postings")
plt.xlabel("Months")
plt.ylabel("Job Count")
plt.show()
```
