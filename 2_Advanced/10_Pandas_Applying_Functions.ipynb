```python
# Load Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Load Data
dataset = load_dataset("lukebarousse/data_jobs")
df = dataset["train"].to_pandas()

# Clean Data
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
```

```python
df_salary = df[pd.notna(df["salary_year_avg"])].copy()

def projected_salary(salary):
    return salary * 1.03

df_salary["salary_year_avg"].apply(projected_salary)
```

```python
# Comparing how the new salary compares to the original salary
df_salary["salary_year_inflated"] = df_salary["salary_year_avg"].apply(projected_salary)

df_salary[["salary_year_avg", "salary_year_inflated"]].head(10)
```

```python
# Using a lambda function to solve the same problem
df_salary["salary_year_inflated"] = df_salary["salary_year_avg"].apply(lambda salary: salary * 1.03)

df_salary[["salary_year_avg", "salary_year_inflated"]].head(10)
```

```python
# Straight forward way
df_salary["salary_year_inflated"] = df_salary["salary_year_avg"] * 1.03

df_salary[["salary_year_avg", "salary_year_inflated"]].head(10)
```

### Calculate projected salary next year
- Senior roles assume 5%
- Other roles assume 3%

```python
df_salary["salary_year_inflated"] = df_salary["salary_year_avg"].apply(lambda salary: salary * 1.03)

df_salary[["salary_year_avg", "salary_year_inflated"]]
```

```python
def projected_salary(row):
    if "Senior" in row["job_title_short"]:
        return 1.05 * row["salary_year_avg"]
    else:
        return 1.03 * row["salary_year_avg"]

df_salary["salary_year_inflated"] = df_salary.apply(projected_salary, axis=1)

df_salary[["job_title_short", "salary_year_avg", "salary_year_inflated"]]
```

```python
# Checking is code worked by filtering for senior roles
df_salary.loc[
    df_salary['job_title_short'].str.startswith("Senior"),
    ["job_title_short", "salary_year_avg", "salary_year_inflated"]
]
```

```python
# Using the lambda function
df_salary["salary_year_inflated"] = df_salary.apply(lambda row: 1.05 * row["salary_year_avg"] if "Senior" in row["job_title_short"] else 1.03 * row["salary_year_avg"], axis=1)

df_salary[["job_title_short", "salary_year_avg", "salary_year_inflated"]]
```
