```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset("lukebarousse/data_jobs")
df = dataset["train"].to_pandas()

# Data Cleaning
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
```

# Count of Job Titles

```python
# Comparing the capabilities if pivot tables and groupby
df.pivot_table(index="job_title_short", aggfunc="size") # the size aggfunc counts all the rows in a group(value/column not required).
```

```python
df.groupby(by="job_title_short").size()
```

```python
df.pivot_table(values="salary_year_avg", index="job_title_short", aggfunc="median")
```

```python
df.groupby(by="job_title_short")["salary_year_avg"].median()
```

```python
# Calculating the median salary of the job titles broken down by country
df.pivot_table(values="salary_year_avg", index="job_country", columns="job_title_short", aggfunc="median")
```

```python
# plotting the top 6 countries with their job_titles according to their median yearly salary
top_countries = df["job_country"].value_counts().head(6).index

df_job_country_salary = df.pivot_table(
    values="salary_year_avg", 
    index="job_country", 
    columns="job_title_short", 
    aggfunc="median"
)

df_job_country_salary = df_job_country_salary.loc[top_countries] # Filtering the top 6 countries

job_titles = ["Data Analyst", "Data Engineer", "Data Scientist"] # Specifying the top 3 jobs titles

df_job_country_salary = df_job_country_salary[job_titles]

```

```python
df_job_country_salary
```

```python
# plotting df_job_country_salary
df_job_country_salary.plot(kind="bar")
plt.ylabel("Median Yearly Salary ($USD)")
plt.xlabel("")
plt.title("Median Salary by Country and Job Title")
plt.xticks(rotation=45, ha="right")
plt.show()
```
