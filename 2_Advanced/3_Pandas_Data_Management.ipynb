### Copy( )

```python
dataset = load_dataset("lukebarousse/data_jobs")
df_original = dataset["train"].to_pandas()
```

```python
# Proving why we use the df.copy() function instead of creating the new variable and equating it to the dataset
df_altered = df_original
```

```python
df_altered.loc[:5, "salary_year_avg"]
```

```python
median_salary = df_altered["salary_year_avg"].median()
median_salary
```

```python
df_altered["salary_year_avg"] = df_altered["salary_year_avg"].fillna(median_salary)
df_altered.loc[:5, "salary_year_avg"]
```

```python
df_original.loc[:5, "salary_year_avg"]
```

```python
# Conclusion: Our orinal and altered end up storing the same data because we equated their variables. But we want to keep the original intact. Hence why we use df.copy()
```

```python
# Proving that by setting the df_original to df_altered, they became the same by using the id() function
print(id(df_original))
print(id(df_altered))
```

```python
# Using the comparision operator to check if they are the same
id(df_original) == id(df_altered)
```

```python
# Using the df.copy method
dataset = load_dataset("lukebarousse/data_jobs")
df_original = dataset["train"].to_pandas()
```

```python
df.altered = df_original.copy()
```

```python
print("ID of df_original: ", id(df_original))
print("ID of df_altered: ", id(df_altered))
print("Are the two dataframes the same? ", id(df_original) == id(df_altered))
```

```python
median_salary = df_altered["salary_year_avg"].median()

df_altered["salary_year_avg"] = df_altered["salary_year_avg"].fillna(median_salary)
df_altered.loc[:5, "salary_year_avg"]
```
